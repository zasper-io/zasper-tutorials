# Live output

## Progress bars

```python
import time
from tqdm.auto import tqdm

for _ in tqdm(range(100), desc="Training"):
    time.sleep(0.03)
```

## Output as it happens

```python
for i in range(10, 0, -1):
    print(f"Launching in {i}…", flush=True)
    time.sleep(0.3)
print("🚀")
```

## Animation

```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import clear_output

x = np.linspace(0, 4 * np.pi, 400)
for phase in np.linspace(0, 2 * np.pi, 40):
    clear_output(wait=True)
    fig, ax = plt.subplots(figsize=(8, 3))
    ax.plot(x, np.sin(x + phase))
    ax.plot(x, np.cos(x - phase), alpha=0.6)
    ax.set_ylim(-1.2, 1.2)
    plt.show()
```

## A firehose

Fifty thousand lines, and none of them dropped.

```python
for i in range(50_000):
    print(f"line {i:>6}: {'#' * (i % 60)}")
```

## Stop me

This runs forever. Interrupt the kernel to stop it.

```python
n = 0
try:
    while True:
        n += 1
        time.sleep(0.1)
except KeyboardInterrupt:
    print(f"Interrupted after {n / 10:.1f} seconds")
```
