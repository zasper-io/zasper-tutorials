## A 3D chart you can rotate

```python
import numpy as np
import plotly.graph_objects as go

x = y = np.linspace(-3, 3, 60)
X, Y = np.meshgrid(x, y)
Z = np.sin(X**2 + Y**2) / (1 + X**2 + Y**2)
go.Figure(go.Surface(x=X, y=Y, z=Z, colorscale="Viridis")).update_layout(height=500)
```
