```r
suppressPackageStartupMessages(library(dplyr))
library(ggplot2)

penguins <- read.csv("../data/penguins.csv") |> na.omit()
head(penguins)
```

```r
penguins |>
  group_by(species) |>
  summarise(across(c(bill_length_mm, flipper_length_mm, body_mass_g), mean), n = n())
```

```r
options(repr.plot.width = 8, repr.plot.height = 5)

ggplot(penguins, aes(flipper_length_mm, body_mass_g, colour = species)) +
  geom_point(alpha = 0.7) +
  geom_smooth(method = "lm", formula = y ~ x, se = FALSE) +
  labs(x = "Flipper length (mm)", y = "Body mass (g)") +
  theme_minimal(base_size = 14)
```

```r
summary(lm(body_mass_g ~ flipper_length_mm + species, data = penguins))
```
