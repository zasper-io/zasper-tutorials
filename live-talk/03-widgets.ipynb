## A penguin explorer

```python
import matplotlib.pyplot as plt
import pandas as pd

penguins = pd.read_csv("../data/penguins.csv").dropna()
numeric = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]

@widgets.interact(x=numeric, y=widgets.Dropdown(options=numeric, value="body_mass_g"))
def explore(x, y):
    fig, ax = plt.subplots(figsize=(7, 4))
    for species, group in penguins.groupby("species"):
        ax.scatter(group[x], group[y], label=species, alpha=0.7)
    ax.set_xlabel(x)
    ax.set_ylabel(y)
    ax.legend()
    plt.show()
```

## Buttons and callbacks

```python
button = widgets.Button(description="Feed a penguin", button_style="info")
fed = widgets.Output()
count = 0

def feed(_):
    global count
    count += 1
    with fed:
        fed.clear_output()
        print(f"{count} penguin{'s' if count > 1 else ''} fed 🐧")

button.on_click(feed)
widgets.VBox([button, fed])
```

Close this tab and open it again: the widgets come back from the running kernel.
