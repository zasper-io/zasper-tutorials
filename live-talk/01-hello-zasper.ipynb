## Some real data

The Palmer Penguins: 344 penguins from three islands in Antarctica.

```python
import pandas as pd

penguins = pd.read_csv("../data/penguins.csv").dropna()
penguins.head()
```

```python
penguins.groupby("species")[["bill_length_mm", "flipper_length_mm", "body_mass_g"]].mean().round(1)
```

## Interactive charts

Hover, zoom and click the legend.

```python
import plotly.express as px

px.scatter(
    penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    color="species",
    size="body_mass_g",
    hover_data=["island", "sex"],
    title="Bill shape separates the three species",
)
```

## And the classics

Matplotlib works too.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
for species, group in penguins.groupby("species"):
    ax.hist(group["body_mass_g"], bins=20, alpha=0.6, label=species)
ax.set_xlabel("Body mass (g)")
ax.set_ylabel("Penguins")
ax.legend()
plt.show()
```
