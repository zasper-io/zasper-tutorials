## Fast loops, no vectorising needed

```julia
function estimate_pi(n)
    inside = 0
    for _ in 1:n
        inside += rand()^2 + rand()^2 <= 1
    end
    4inside / n
end

estimate_pi(10)  # compile once
@time estimate_pi(100_000_000)
```

## The same penguins

```julia
using Statistics

rows = split.(readlines("../data/penguins.csv")[2:end], ",")
rows = filter(r -> r[6] != "NA", rows)
mass = Dict(String(s) => mean(parse(Float64, r[6]) for r in rows if r[1] == s)
            for s in unique(first.(rows)))
```

## Types that draw themselves

```julia
struct BarChart
    data::Dict{String,Float64}
end

function Base.show(io::IO, ::MIME"image/svg+xml", c::BarChart)
    names = sort(collect(keys(c.data)))
    top = maximum(values(c.data))
    print(io, """<svg xmlns="http://www.w3.org/2000/svg" width="420" height="$(40length(names) + 10)">""")
    for (i, name) in enumerate(names)
        y = 40(i - 1) + 10
        w = 300 * c.data[name] / top
        print(io, """<text x="0" y="$(y + 20)" font-size="14" fill="#64748b">$name</text>""")
        print(io, """<rect x="80" y="$y" width="$w" height="28" rx="4" fill="#9558b2"/>""")
        print(io, """<text x="$(w + 88)" y="$(y + 20)" font-size="13" fill="#64748b">$(round(Int, c.data[name])) g</text>""")
    end
    print(io, "</svg>")
end

BarChart(mass)
```
