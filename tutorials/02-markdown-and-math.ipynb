## Maths

Write LaTeX between single dollar signs for inline maths, like $a^2 + b^2 = c^2$, and between
double dollar signs for a display equation:

$$
\hat{\beta} = (X^\top X)^{-1} X^\top y
$$

Jupyter's bracket delimiters work too: \(E = mc^2\) inline, and

\[
f(x) = \frac{1}{\sigma\sqrt{2\pi}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
\]

## Markdown from code

A code cell can produce Markdown and maths as its output.

````python
from IPython.display import Markdown

rows = "\n".join(f"| {n} | ${n}^2 = {n * n}$ |" for n in range(1, 6))
Markdown("| n | square |\n|---|---|\n" + rows)
````
