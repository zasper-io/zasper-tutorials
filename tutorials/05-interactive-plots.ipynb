# Interactive plots with Plotly

Plotly charts are interactive in Zasper: hover for values, drag to zoom, double-click to reset
and click legend entries to hide series.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

penguins = pd.read_csv("../data/penguins.csv").dropna()
```

## Scatter

```python
px.scatter(
    penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    color="species",
    symbol="sex",
    hover_data=["island"],
    labels={"flipper_length_mm": "Flipper length (mm)", "body_mass_g": "Body mass (g)"},
)
```

## Distributions

```python
px.histogram(penguins, x="body_mass_g", color="species", marginal="box", barmode="overlay", opacity=0.7)
```

```python
px.violin(penguins, y="flipper_length_mm", x="species", color="sex", box=True, points="all")
```

## Small multiples

```python
px.scatter(
    penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    color="species",
    facet_col="island",
    facet_row="sex",
    height=550,
)
```

## Many dimensions at once

```python
px.scatter_matrix(
    penguins,
    dimensions=["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"],
    color="species",
    height=650,
).update_traces(diagonal_visible=False, marker={"size": 4})
```

```python
px.scatter_3d(penguins, x="bill_length_mm", y="bill_depth_mm", z="flipper_length_mm", color="species", height=550)
```

## Animation

Plotly ships the Gapminder dataset. Press play.

```python
gapminder = px.data.gapminder()
px.scatter(
    gapminder,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    animation_frame="year",
    log_x=True,
    size_max=55,
    range_x=[100, 100_000],
    range_y=[25, 90],
)
```

## Maps

```python
px.choropleth(
    gapminder.query("year == 2007"),
    locations="iso_alpha",
    color="lifeExp",
    hover_name="country",
    color_continuous_scale="Viridis",
)
```

## Full control with graph objects

`plotly.express` builds figures from dataframes. `plotly.graph_objects` builds them trace by
trace, when you need something custom.

```python
means = penguins.groupby("species")[["bill_length_mm", "bill_depth_mm", "flipper_length_mm"]].mean()
normalised = means / means.max()

fig = go.Figure()
for species, row in normalised.iterrows():
    fig.add_trace(go.Scatterpolar(r=list(row) + [row.iloc[0]], theta=list(row.index) + [row.index[0]], name=species, fill="toself", opacity=0.6))
fig.update_layout(polar={"radialaxis": {"range": [0.5, 1]}}, title="Species profiles, relative to the largest")
```
