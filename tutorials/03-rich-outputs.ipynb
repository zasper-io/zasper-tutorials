# Rich outputs

A cell's output is not just text. The kernel sends a **MIME bundle**: the same result in several
formats, such as `text/html` and `text/plain`, and Zasper shows the richest one it can.

Zasper renders `text/plain`, `text/html`, `text/markdown`, `text/latex`, `image/png`,
`image/jpeg`, `image/svg+xml`, `application/json`, Plotly figures and Jupyter widgets.

## The last expression is the result

A cell's last expression becomes its result. Other values need `display()`.

```python
from IPython.display import HTML, JSON, Image, Latex, Markdown, SVG, display

display("displayed first")
"the cell's result"
```

## Images

```python
import io

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6, 3))
ax.imshow(np.random.default_rng(0).random((20, 60)), cmap="magma")
ax.axis("off")
buffer = io.BytesIO()
fig.savefig(buffer, format="png", bbox_inches="tight")
plt.close(fig)

Image(buffer.getvalue())
```

Matplotlib figures appear on their own at the end of a cell, as PNG by default. Ask for SVG to get
sharp vector output:

```python
%config InlineBackend.figure_formats = ["svg"]

x = np.linspace(0, 10, 200)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, np.sin(x) * np.exp(-x / 5))
plt.show()

%config InlineBackend.figure_formats = ["png"]
```

## Your own objects

Give a class a `_repr_html_` (or `_repr_markdown_`, `_repr_svg_`, `_repr_latex_`, …) method and
notebooks will use it whenever an instance is displayed.

```python
class Swatch:
    def __init__(self, *colours):
        self.colours = colours

    def _repr_html_(self):
        chips = "".join(
            f'<div title="{c}" style="width:48px;height:48px;border-radius:6px;background:{c}"></div>'
            for c in self.colours
        )
        return f'<div style="display:flex;gap:6px">{chips}</div>'

    def __repr__(self):
        return f"Swatch{self.colours}"

Swatch("#0f766e", "#14b8a6", "#f59e0b", "#ef4444", "#6366f1")
```

## One bundle, several formats

`display(..., raw=True)` sends a MIME bundle directly. Zasper picks the richest format it knows.

```python
display(
    {
        "text/plain": "Plain text, for terminals",
        "text/html": "<b style='color:#0f766e'>HTML, for notebooks</b>",
    },
    raw=True,
)
```
