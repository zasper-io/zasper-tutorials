# Code intelligence

Zasper asks two sources for help while you type:

- **The kernel** knows what exists *right now*: the variables you have defined and the attributes of
  real objects.
- **A language server** reads the notebook's *source*: it knows about names in cells you have not
  run yet, and it spots mistakes before you run anything.

This folder's environment includes the `basedpyright` language server. Open the `zasper-tutorials`
folder in Zasper and trust it: Zasper looks on your `PATH` and then in the folder's `.venv`, so it
finds the server with no setup. The language server's status is shown in the status bar.

## Names from cells you have not run

The kernel has never seen `wingspan_ratio` below, because the cell has not run. The language
server has read it, though. Type `wingspan` in the empty cell after it and the completion comes
from the language server.

```python
def wingspan_ratio(flipper_mm: float, mass_g: float) -> float:
    """Flipper length per kilogram of body mass."""
    return flipper_mm / (mass_g / 1000)
```

## Problems

The language server underlines mistakes as you type. Type checking is `off` by default, which still
underlines names that are not defined and imports that cannot be found. To see one, add a new cell
and type `describe_penguin(penguins)`: the misspelled name is underlined before you run it.

The code cell under this text has two mistakes of another kind: a misspelled attribute, and a
string passed where a number is expected. These are type errors, so with Type checking `off` nothing
is underlined. Open **Settings** and set **Type checking**, under **Language servers**, to `basic`.
The underlines appear at once. Hover them to read them, then fix them.

```python
heaviest = max(penguins, key=lambda p: p.mass_g)
print(heaviest.specis)

wingspan_ratio(heaviest.flipper_mm, "5000")
```

## Formatting

or run **Format Notebook** from the command palette to format every cell.

```python
def  summarise( items,key ):
    totals={ }
    for item in items :
        totals[ getattr(item,key) ]=totals.get(getattr(item,key),0)+1
    return dict(sorted(totals.items(),key=lambda kv:-kv[1]))
summarise(penguins,'species')
```
