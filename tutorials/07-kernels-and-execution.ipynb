# Kernels and execution

The **kernel** is the process that runs your code. Zasper starts one for each open notebook and
talks to it over the Jupyter protocol, so any language with a Jupyter kernel works, not just
Python. This tutorial covers what the kernel does and how to control it.

## Which kernel?

The kernel's name is shown in the notebook toolbar. Click it, or run **Change Kernel** from the
command palette, to switch to another installed kernel or Python environment.

```python
import os
import sys

print("Interpreter:", sys.executable)
print("Process ID: ", os.getpid())
print("Directory:  ", os.getcwd())
```

The kernel's working directory is the notebook's folder, which is why these notebooks can read
`../data/penguins.csv`.

## Execution order

Cells run in the order you run them, not the order they appear. Run the next three cells, then
run the middle one again.

```python
total = 0
```

```python
total += 10
total
```

```python
print(f"total is {total}")
```

Each run of the middle cell adds 10 more. The execution counts in the prompts show what happened.
Before sharing a notebook, use **Restart Kernel and Run All Cells** to check it works from top to
bottom.

## Output while it runs

Output streams to the page as the kernel produces it.

```python
import time

for step in ["Loading data", "Cleaning", "Fitting model", "Evaluating"]:
    print(f"{step}…", flush=True)
    time.sleep(0.6)
print("Done")
```

```python
from tqdm.auto import tqdm

for _ in tqdm(range(50), desc="Processing"):
    time.sleep(0.04)
```

`clear_output(wait=True)` replaces a cell's output in place, which is how you make a live
display.

```python
from datetime import datetime

from IPython.display import clear_output

for _ in range(10):
    clear_output(wait=True)
    print(datetime.now().strftime("%H:%M:%S"), flush=True)
    time.sleep(0.5)
```

Zasper does not throttle output. A cell that prints a lot shows all of it.

```python
for i in range(20_000):
    print(i)
```

Long output is folded to keep the notebook readable. Use **Expand or Collapse Output** from the
command palette to see all of it.

## Interrupting

The next cell never finishes on its own. Run it, then stop it with **Interrupt Kernel** from the
toolbar or the command palette. Interrupting raises `KeyboardInterrupt` in your code, which you can
catch to clean up.

```python
seconds = 0
try:
    while True:
        time.sleep(1)
        seconds += 1
        print(f"Still running after {seconds} s", flush=True)
except KeyboardInterrupt:
    print("Stopped cleanly")
```

## Restarting

**Restart Kernel** stops the kernel process and starts a fresh one. Every variable is lost. Run
this cell, restart the kernel, then run the one after it.

```python
precious = "important result"
os.getpid()
```

```python
precious
```

After a restart, that cell fails with `NameError`, and `os` is gone too: rerun the imports first.
A kernel that dies, for example by running out of memory, is restarted the same way.
