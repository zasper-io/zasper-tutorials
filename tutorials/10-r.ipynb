## Data frames

IRkernel shows data frames as HTML tables.

```r
penguins <- read.csv("../data/penguins.csv", stringsAsFactors = TRUE)
str(penguins)
```

```r
summary(penguins)
```

```r
penguins <- na.omit(penguins)
head(penguins, 10)
```

## dplyr

```r
suppressPackageStartupMessages(library(dplyr))

penguins |>
  group_by(species, sex) |>
  summarise(
    n = n(),
    mass = mean(body_mass_g),
    flipper = mean(flipper_length_mm),
    .groups = "drop"
  ) |>
  arrange(desc(mass))
```

## ggplot2

```r
library(ggplot2)
options(repr.plot.width = 9, repr.plot.height = 5)

ggplot(penguins, aes(bill_length_mm, bill_depth_mm, colour = species)) +
  geom_point(alpha = 0.7, size = 2) +
  labs(x = "Bill length (mm)", y = "Bill depth (mm)", colour = "Species") +
  theme_minimal(base_size = 14)
```

```r
ggplot(penguins, aes(body_mass_g, fill = species)) +
  geom_density(alpha = 0.5) +
  facet_wrap(~ island) +
  theme_minimal(base_size = 13)
```

## Models

```r
model <- lm(body_mass_g ~ flipper_length_mm + sex + species, data = penguins)
summary(model)
```

```r
anova(model)
```
