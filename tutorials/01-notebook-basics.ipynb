# Notebook basics

A notebook is a list of **cells**. Code cells run on a **kernel**, a separate process that
keeps your variables between runs. Markdown cells hold prose, like this one.

This tutorial covers moving around, running code and editing cells. Work through it from top to
bottom and run each cell as you go.

## Running a cell

Click in the cell below and press **Shift+Enter**. That runs the cell and moves to the next one.
**Ctrl+Enter** runs the cell and stays where you are.

```python
1 + 1
```

The number that appears in the prompt beside a cell, like `[1]`, is its **execution count**. It
records the order in which cells ran, which is not always their order on the page.

## Two modes

Zasper follows Jupyter's two keyboard modes:

| Mode | What has the keyboard | How to get there |
|---|---|---|
| **Edit mode** | The cell's text editor. You type code or text. | **Enter**, or click inside the cell |
| **Command mode** | The notebook. Keys act on whole cells. | **Escape** |

In command mode, **↑** and **↓** move between cells. Try it: press **Escape**, move up and down
a few cells, then press **Enter** to start editing one.

## Working with cells

| Action | Shortcut |
|---|---|
| Insert a cell above | **Ctrl+Shift+A** |
| Insert a cell below | **Ctrl+Shift+B** |
| Move a cell up or down | **Ctrl+Shift+↑** / **Ctrl+Shift+↓** |
| Delete a cell | **Ctrl+Shift+D** |
| Undo a cell operation | **Cmd+Shift+Z** (Ctrl+Shift+Z elsewhere) |
| Turn a cell into Markdown | **Ctrl+Shift+M** |
| Turn a cell into code | **Ctrl+Shift+Y** |
| Save | **Cmd+S** (Ctrl+S elsewhere) |

Every other notebook action is in the **command palette**: **Cmd+Shift+P** (Ctrl+Shift+P
elsewhere). Type "notebook" to see them all. **F1** lists every keyboard shortcut.

## The kernel remembers

Variables live in the kernel, not in the page. Run these two cells in order.

```python
greeting = "Hello from the kernel"
```

```python
greeting.upper()
```

Now change the first cell to a different greeting, but only run the *second* one. It still shows
the old value, because the kernel has not seen your edit. When a notebook behaves strangely, the
reliable fix is **Restart Kernel and Run All Cells** from the command palette.

## Find and replace

**Cmd+F** (Ctrl+F elsewhere) searches every cell in the notebook, code and Markdown alike, and can
replace matches too. Try searching for `kernel`.

## Exporting

To share a notebook with someone who does not use Jupyter, export it from the command palette:

- **Export as an HTML Page**: a single file, with outputs, that opens in any browser.
- **Export as Markdown**: for docs sites and READMEs.
- **Export as a Script**: the code cells only, as a `.py` file.
